```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest")
print("Model is ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate

user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Which is the capital city of {country}?")
])

user_prompt = user_chat_prompt_template.invoke({"country": "France"})
response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
from langchain_core.prompts import ChatPromptTemplate

user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Please explain the {concept} in simple terms."),
    ("system", "You are an expert mathematician and teacher. Please provide a clear and concise explanation of the concept.")
])

user_prompt = user_chat_prompt_template.invoke({"concept": "pythagoras theorem"})
response = model.invoke(user_prompt)
print(type(response))
print(response)
```
