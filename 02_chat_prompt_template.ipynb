```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest")
print("Model is ready")
```

```python
from langchain_core.prompts import ChatPromptTemplate

user_chat_prompt_template = ChatPromptTemplate([
    ("user", "Which is the capital city of {country}?")
])

user_prompt = user_chat_prompt_template.invoke({"country": "France"})
response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
# Explanation of conversational state
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

user_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{user_input}")
])

chat_history = [
    HumanMessage(content="Which is the capital city of France?"),
    AIMessage(content="The capital city of France is Paris.")
]

user_prompt = user_prompt_template.format_messages(chat_history=chat_history, 
                                                   user_input="Which is the famous landmark there?")

response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
chat_history = []
while True:
    question = input("Ask a question (or type 'exit' to quit): ")
    if question.lower() == 'quit':
        print("Exiting the chat.")
        break

    user_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    MessagesPlaceholder(variable_name="chat_history"),
    ("human", "{user_input}")
    ])
    #chat_history.append(SystemMessage("You are a helpful AI assistant."))
    chat_history.append(HumanMessage(content=question))
    user_prompt = user_prompt_template.format_messages(chat_history=chat_history, 
                                                   user_input=question)
    response = model.invoke(user_prompt)
    print("AI Response:", response.content)
    chat_history.append(SystemMessage(content=response.content))

print(chat_history)
```
