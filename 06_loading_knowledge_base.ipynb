```python
# Example - Knowledgebase in text format
from langchain_community.document_loaders import TextLoader
loader = TextLoader("Gen_AI.txt")
text_pages = loader.load()
print(f'No of pages: {len(text_pages)}')
```

```python
# Example - Knowledgebase in pdf format
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("Python_Programming.pdf")
pdf_pages = loader.load()
print(f'No of pages: {len(pdf_pages)}')
```

```python
# Example - Knowledgebase in csv format
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("employees.csv")
csv_pages = loader.load()
print(f'No of pages: {len(csv_pages)}')
```

```python
# Example - Knowledgebase in web document format
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader("https://en.wikipedia.org/wiki/Generative_AI")
web_pages = loader.load()
print(f'No of pages: {len(web_pages)}')
```

```python
# Complete knowledgebase
all_pages = text_pages + pdf_pages + csv_pages + web_pages
```
