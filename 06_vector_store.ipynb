```python
from langchain_ollama import OllamaEmbeddings

embeddings = OllamaEmbeddings(model="nomic-embed-text")
print('Embedding model is ready')
```

```python
# Example - creating simple vector store in memory
vector_store = {}
input_text = "Zensar allows 18 leaves per year"
words = input_text.split()
for word in words:
    embedding = embeddings.embed_query(word)
    vector_store[word] = embedding
print(f'Vector store: {vector_store}')
```

```python
# Example - Store the vectors using FAISS library
from langchain_community.vectorstores import FAISS
import faiss

words = ["Zensar", "offers", "18", "leaves", "per", "annum"]
vectors = FAISS.from_texts(words, embeddings)
vectors.save_local("faiss_zen_leave_policy")
print(vectors)
```

```python
# Example - Store the vectors using Chroma database
from langchain_community.vectorstores import Chroma

words = ["Zensar", "offers", "18", "leaves", "per", "annum"]

vector_store = Chroma.from_texts(words, embeddings)
print(f'Vector store using Chroma {vector_store}')

persist_directory = "chroma_db"

# Create a Chroma vector store from the text chunks and embeddings, and persist it to disk
vector_store = Chroma.from_texts(
    texts=words,
    embedding=embeddings,
    persist_directory=persist_directory
)

print('Chroma vector store is saved locally!!')
```
