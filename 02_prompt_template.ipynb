```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="llama3.2:latest")
print("Model is ready")
```

```python
# Example of using a prompt template with LangChain having one variable
from langchain_core.prompts import PromptTemplate

user_prompt_template = PromptTemplate(
    template="Which is the capital city of {country}?",
    input_variables=["country"]
)
user_prompt = user_prompt_template.format(country="India")

response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
# Example of using a prompt template with LangChain having Mustache variables

input = {
    "name": "Sachin Tendulkar",
    "team": "Indian cricket team"
}

user_prompt_template = PromptTemplate(
    template="Please find information about {name} working with {team}.",
    input_variables=["name", "team"]
)

#user_prompt = user_prompt_template.format(name=input["name"], team=input["team"])
user_prompt = user_prompt_template.format(**input) # Mustache variables

response = model.invoke(user_prompt)
print(type(response))
print(response)
```
