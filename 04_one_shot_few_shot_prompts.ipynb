```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print("Model is ready")
```

```python
# Example - One shot learning
from langchain_core.prompts import PromptTemplate

one_shot_prompt = (
    f"Example:\n"
    f"Q: Translate 'Hello, how are you? into Hindi'"
    f"A: Aap kaise ho?\n\n"
    f"Now answer this:\n"
    f"Q: Translate 'What is your name?' into Hindi\n"
    f"A:"
    )

user_prompt_template = PromptTemplate(
    template=one_shot_prompt
)

user_prompt = user_prompt_template.format()

response = model.invoke(user_prompt)
print(type(response))
print(response)
```

```python
# Example - Few shot learning
from langchain_core.prompts import PromptTemplate

few_shot_prompt = """
Examples:
Message: My laptop hangs frequently
Category: Laptop

Message: Food quality is poor in office canteen
Category: Food

Classify the following message into category:
Message: Bus driver is a rash driver
Category:
"""

user_prompt_template = PromptTemplate(
    template=few_shot_prompt
)

user_prompt = user_prompt_template.format()

response = model.invoke(user_prompt)
print(type(response))
print(response)
```
