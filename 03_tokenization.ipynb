```python
import tiktoken

encoding = tiktoken.get_encoding("cl100k_base")
#prompt = "I work in IT industry and I like work in music also"
prompt = "Generative AI is a type of artificial intelligence that can create new content."

tokens = encoding.encode(prompt)
print(len(tokens))
print(f"decoded tokens: {encoding.decode(tokens)}")

for token in tokens:
    print(f"{encoding.decode([token])} - {token}")
```
